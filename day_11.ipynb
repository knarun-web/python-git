```python
# Parameters
INPUT_FOLDER = "data"
```

## Task 4 — Read orders CSV and print row count

```python
import csv
import os

filepath = os.path.join(INPUT_FOLDER, "orders.csv")

with open(filepath, newline="", encoding="utf-8") as f:
    reader = csv.reader(f)
    rows = list(reader)

print(f"Row count (including header): {len(rows)}")
print(f"Data rows: {len(rows) - 1}")
```

## Task 5 — Print column names and first 5 rows

```python
headers = rows[0]
print("Columns:", headers)
print("\nFirst 5 rows:")
for row in rows[1:6]:
    print(row)
```

## Task 6 — Count missing values per column

```python
missing = {col: 0 for col in headers}

for row in rows[1:]:
    for col, val in zip(headers, row):
        if val.strip() == "":
            missing[col] += 1

print("Missing values per column (orders.csv):")
for col, count in missing.items():
    print(f"  {col}: {count}")
```

## Task 7 — Summary for all Olist CSV files

```python
csv_files = sorted([f for f in os.listdir(INPUT_FOLDER) if f.endswith(".csv")])

for filename in csv_files:
    fp = os.path.join(INPUT_FOLDER, filename)
    with open(fp, newline="", encoding="utf-8") as f:
        reader = csv.reader(f)
        file_rows = list(reader)
    file_headers = file_rows[0] if file_rows else []
    print(f"\n--- {filename} ---")
    print(f"  Columns ({len(file_headers)}): {file_headers}")
    print(f"  Data rows: {len(file_rows) - 1}")
```

## Task 10 — Handle both comma and semicolon delimiters

```python
# Create a semicolon-delimited copy of orders.csv (if not already present)
src = os.path.join(INPUT_FOLDER, "orders.csv")
dst = os.path.join(INPUT_FOLDER, "ordersdelimiter.csv")

with open(src, newline="", encoding="utf-8") as fin, \
     open(dst, "w", newline="", encoding="utf-8") as fout:
    reader = csv.reader(fin)
    writer = csv.writer(fout, delimiter=";")
    for row in reader:
        writer.writerow(row)

print(f"Semicolon-delimited file created: {dst}")
```

```python
# Auto-detect delimiter and read both files
def read_csv_auto(filepath):
    with open(filepath, newline="", encoding="utf-8") as f:
        sample = f.read(2048)
        dialect = csv.Sniffer().sniff(sample)
        f.seek(0)
        reader = csv.reader(f, dialect)
        file_rows = list(reader)
    return file_rows, dialect.delimiter

for fname in ["orders.csv", "ordersdelimiter.csv"]:
    r, delim = read_csv_auto(os.path.join(INPUT_FOLDER, fname))
    print(f"{fname}: {len(r) - 1} data rows, delimiter detected as {repr(delim)}")
```
